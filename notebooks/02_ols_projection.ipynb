```python
from pathlib import Path
import sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'course.json').is_file())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from study.support import load_answers, attempt, npv
a = load_answers()
```

```python
X=np.column_stack([np.ones(5),np.arange(5)])
y=np.array([1.,2.,2.,4.,5.])
print('X shape:',X.shape,'y shape:',y.shape)
# Вставь собственный расчёт X.T@X и X.T@y после ручного вывода.
out=attempt(a.ols,X,y)
if out is not None:
    print(out)
    print('orthogonality:',X.T@out['residuals'])
```

## Мои предпосылки
[Раздели алгебру оценки, стохастическую модель,

```python
if out is not None:
    fitted=X@out['beta']
    fig,ax=plt.subplots(figsize=(6,3))
    ax.scatter(X[:,1],y,label='synthetic observations')
    ax.plot(X[:,1],fitted,label='linear projection')
    ax.set(xlabel='x',ylabel='y');ax.legend();plt.show()
    scaled=X.copy();scaled[:,1]*=10
    other=attempt(a.ols,scaled,y)
    if other is not None:
        print('same predictions:',np.allclose(scaled@other['beta'],fitted))
```
