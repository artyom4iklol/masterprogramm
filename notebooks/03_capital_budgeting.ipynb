```python
revenue=np.array([100.,110.,121.])
cash_cost=np.array([60.,66.,72.6])
cf=attempt(a.project_cashflows,120,revenue,cash_cost,.25,.1,30)
if cf is not None:
    display(pd.DataFrame({'time':range(4),'FCFF':cf}))
    for r in [0,.1,.2]: print('rate=',r,'NPV=',npv(cf,r))
```

## Моя временная линия и мост от прибыли к FCFF

```python
if cf is not None:
    rates=np.linspace(0,.2,21)
    multipliers=np.linspace(.8,1.2,21)
    values=np.array([[npv(a.project_cashflows(120,revenue*z,cash_cost,.25,.1,30),r)
                      for r in rates] for z in multipliers])
    fig,ax=plt.subplots(figsize=(7,4))
    im=ax.imshow(values,origin='lower',aspect='auto',extent=[0,20,.8,1.2])
    ax.set(xlabel='discount rate, percent',ylabel='revenue multiplier',title='Scenario NPV')
    fig.colorbar(im,ax=ax,label='cash units');plt.show()
    # Самостоятельно найди порог NPV=0 и интерпретируй чувствительность.
```
